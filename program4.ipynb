```python
import csv

# Read dataset from the user
filename = input("Enter CSV file name: ")

with open(filename, 'r') as file:
    data = list(csv.reader(file))

# Separate header and examples
header = data[0]
examples = data[1:]

attributes = header[:-1]
n = len(attributes)

# Most specific hypothesis
S = ['Ø'] * n

# Find-S algorithm
for row in examples:
    x = row[:-1]
    label = row[-1].strip().lower()

    # Find-S processes only positive examples
    if label == "yes":
        for i in range(n):
            if S[i] == 'Ø':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

# Display results
print("\nAttributes:")
print(attributes)

print("\nFinal Hypothesis after processing all positive examples:")
print(S)
```
