```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

df = pd.read_csv('nirf_university_rankings.csv')

print("--- Dataset Basic Information ---")
print(df.info())
print("\n--- First 5 Rows ---")
print(df.head())

print("\n--- Summary Statistics ---")
print(df.describe())

print("\n--- Missing Values Count ---")
print(df.isnull().sum())

plt.figure(figsize=(9, 5))
sns.histplot(df['Score'], kde=True, bins=15, color='teal')
plt.title('Distribution of NIRF Scores Across Institutes')
plt.xlabel('Score')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 5))
state_counts = df['State'].value_counts().head(10)
sns.barplot(x=state_counts.values, y=state_counts.index, hue=state_counts.index, palette='viridis', legend=False)
plt.title('Top 10 Indian States by Number of Ranked Institutes')
plt.xlabel('Number of Institutes')
plt.ylabel('State')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

plt.figure(figsize=(7, 5))
sns.scatterplot(x='Rank', y='Score', data=df, color='crimson', alpha=0.8)
plt.title('Correlation: Rank vs Score')
plt.xlabel('Rank')
plt.ylabel('Score')
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()

print('\n--- Correlation between Rank and Score ---')
correlation = df['Rank'].corr(df['Score'])
print(f"Pearson correlation coefficient between Rank and Score: {correlation:.4f}")

correlation_matrix = df[['Rank', 'Score']].corr()

plt.figure(figsize=(6, 5))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Matrix of Rank and Score')
plt.tight_layout()
plt.show()
```
